# 2) arange()
## arange() is used for generating range of values with equal interval of values.
## syntax: varname = numpy.arange(start)
## syntax: varname = numpy.arange(start,stop)
## syntax: varname = numpy.arange(start,stop,step)
## arange() by default creates an object of ndarray with 1D array only.
## To build 2D and ND arrays from arange() function, we use reshape() or shape() attribute.

```python
a = np.arange(10)
print(a,type(a))
```

```python
print("----------------")
print("Dimension=",a.ndim)
print("Shape=",a.shape)
print("Size=",a.size)
print("Internal DT=",a.dtype)
```

```python
a.shape = (2,5)
a
```

```python
a.reshape(5,2)
a
```

```python
a = np.arange(10,22)
print(a,type(a))
print("----------------")
print("Dimension=",a.ndim)
print("Shape=",a.shape)
print("Size=",a.size)
print("Internal DT=",a.dtype)
```

```python
a.shape = (4,3)
print(a)
print("----------------")
print("Dimension=",a.ndim)
print("Shape=",a.shape)
print("Size=",a.size)
print("Internal DT=",a.dtype)
```

```python
a = np.arange(10,22).reshape(2,3,2)
print(a)
print("----------------")
print("Dimension=",a.ndim)
print("Shape=",a.shape)
print("Size=",a.size)
print("Internal DT=",a.dtype)
```

# hstack() & vstack() & dstack()

```python
import numpy as np
```

```python
a = np.array([10,20,30,40])
b = np.array([100,200,300,400])
c = np.hstack((a,b))
print(c,type(c))
```

```python
a = np.array([102,203,320,140])
b = np.array([100,200,300,])
c = np.hstack((a,b))
print(c,type(c))
```

```python
a = np.array([[10,20,30],[40,50,60],[70,80,90]])
b = np.array([[200,400],[300,500],[100,600]])
c = np.hstack((a,b))
print(c,type(c))
```

```python
a = np.array([10,20,30,40])
b = np.array([100,200,300,400])
c = np.vstack((a,b))
print(c,type(c))
```

```python
a = np.array([[10,20,30],[40,50,60],[70,80,90]])
b = np.array([[200,400],[300,500],[100,600]])
c = np.vstack((a,b))
print(c,type(c))
```
