## Indexing is the process of obtaining single value from given iterable object(contains multiple values).
## The values of iterable object contains +ve and -ve indices.
## Now an object of ndarray is also one of the iterable object.
## The element of ndarray contains +ve and -ve indices.
## In NumPy programming, we can apply indexing operation on 3-types of arrays. They are:
- 1) 1-D Indexing
- 2) 2-D Indexing
- 3) N-D Indexing

## 1-D indexing:
## syntax: 1darray[Index]

```python
import numpy as np
a = np.array([10,20,30,40,50,60])
print(a,type(a))
```

```python
a[2]
```

```python
a[4]
```

```python
a[-1]
```

```python
a[-3]
```

```python
len(a)
```

```python
a[-len(a)]
```

```python
## 2D-Indexing
## syntax: 2Darray = [rowindex][colindex]
```

```python
a = np.arange(16).reshape(4,4)
print(a,type(a))
```

```python
a[1,1]
```

```python
a[3,3]
```

```python
a[1][3]
```

```python
a[-2][-2]
```

```python
a[-2,0]
```

# ND - indexing
## syntax: ndarrayobject[MatrixIndex,RowIndex,ColIndex]
## syntax: ndarrayobject[MatrixIndex],[RowIndex],[ColIndex]

```python
a = np.arange(27).reshape(3,3,3)
print(a,type(a))
```

```python
a[0][0][0]
```

```python
a[2][2][2]
```

```python
a[0,1,1]
```

```python
a[-1]
```

# NumPy-Slicing

## The process of obtaining range of values or sub-values from an iterable object is called Slicing.
## We can apply slicing operation on 3-types of ndarray. they are
- 1-D array
- 2-D array
- N-D array

## 1D array:
## syntax: 1darray[Begin:End]
## 1darray[Begin:]
## 1darray[:End] 
## 1darray[Begin:End:step]
## 1darray[:]

```python
a = np.array([10,20,30,40,50,60,70,80,90,15,25,35])
print(a,type(a))
```

```python
a[0:4]
```

```python
a[4:9]
```

```python
a[-4:-1]
```

```python
a[4:]
```

```python
a[9:]
```

```python
a[:5]
```

```python
a[:-5]
```

```python
a[1:12:2]
```

```python
a[::]
```

```python
a[1::2]
```

## 2D-array
## syntax: 2darray[RowBegin:RowEnd:RowStep,ColBegin:ColEnd:ColumnStep]

```python
import numpy as np
```

```python
a = np.arange(16).reshape(4,4)
print(a,type(a))
```

```python
a[0:4,0:1]
```

```python
a[:,0:1]
```

```python
a
```

```python
a[1:3,0:2]
```

```python
a[1:4,1:4]
```

```python
a[1:3,0:4]
```

```python
a[0:4:3,0:4:3]
```
