## Array splitting in numpy like a slice of cake.
## Think of each element in NumPy array as a slice of cake.
## Splitting divides the cake into smaller chunks.
## We can split horizontally, vertically or even diagonally depending on our needs.
- split()
- array_split()
- hsplit()
- vsplit()
- dsplit()

# split()

```python
import numpy as np
```

```python
a = np.arange(6)
print(a)
print("-------------------")
b = np.split(a,2)
print(b,type(b))
```

```python
a = np.arange(6)
print(a)
print("-------------------")
b = np.split(a,3)
print(b,type(b))
```

```python
a = np.arange(10)
print(a)
print("-------------------")
b = np.split(a,2)
print(b,type(b))
```

```python
a = np.arange(10)
print(a)
print("-------------------")
b = np.split(a,5)
print(b,type(b))
```

```python
for val in b:
    print(val,type(val))
```

```python
a = np.arange(8)
b = np.split(a,[3,5,6,10])
print(b)
```

# array_split()

```python
a = np.arange(15)
print(a)
print("-------------------")
b = np.array_split(a,3)
print(b,type(b))
```

```python
a = np.arange(9)
print(a)
print("-------------------")
b = np.array_split(a,5)
print(b,type(b))
```

```python
a = np.arange(10)
print(a)
print("-------------------")
b = np.array_split(a,4)
print(b,type(b))
```

```python
a = np.arange(13)
print(a)
print("-------------------")
b = np.array_split(a,[4,5,7,9])
print(b,type(b))
```

# split()

```python
a = np.array([[3,2,1],[8,9,7],[4,5,6]])
b = np.split(a,3,axis=1)
print(a)
```

```python
a = np.array([[3,2,1],[8,9,7],[4,5,6]])
b = np.split(a,3,axis=0)
print(a)
```

```python
a = np.array([[3,2,1],[8,9,7],[4,5,6]])
b = np.split(a,3,axis=1)
print(a)
```

# vsplit()

```python
a = np.array([[3,2,1],[8,9,7],[4,5,6]])
b = np.vsplit(a,3)
print(a)
```

# hsplit()

```python
a = np.array([[3,2,1,0],[8,1,9,7],[4,4,5,6]])
b = np.hsplit(a,2)
print(a)
```

```python
a = np.array([[3,2,1],[1,9,7],[4,5,6]])
b = np.hsplit(a,3)
print(a)
```

# dsplit()

```python
a = np.arange(24).reshape((2,3,4))
b = np.dsplit(a,2)
print(b,type(b))
```
